```python
import pandas as pd

# 1. load data

df = pd.read_csv("../data/duolingo_raw.csv.gz", compression="gzip")
df.shape
```

```python
# cleaning even though the fraction is small
df = df.drop_duplicates()
df.shape
```

```python
df["learning_language"].value_counts()
```

```python
#average recall rate for each language
df.groupby("learning_language")["p_recall"].mean().sort_values(ascending=False)
```

```python
# check the spread, not just the average
df.groupby("learning_language")["p_recall"].describe()
```

```python
# filtering to only French = fr and German = de
df_fr_de = df[df["learning_language"].isin(["fr", "de"])]
df_fr_de.shape
```

filters cleaned dataframe (the one with duplicates already dropped) down to just French and German rows, and shows the resulting row/column count should be somewhere around 3.3 million rows, matching what we calculated from the value counts earlier.
